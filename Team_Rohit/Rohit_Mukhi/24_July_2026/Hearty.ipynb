```python
import os
import subprocess
import time
from pyngrok import ngrok
from google.colab import userdata

PORT = 8890

# 1. Start Ngrok Tunnel
token = userdata.get('NGROK_TOKEN')
ngrok.set_auth_token(token)
ngrok.kill()

public_url = ngrok.connect(PORT, proto="http")
print(f"🌍 Public URL: {public_url}")
print("👉 Open the URL above. On first launch, Unsloth Studio will ask you to set your password in the UI.\n")

# 2. Launch Unsloth Studio Subprocess
unsloth_path = "/root/.unsloth/studio/unsloth_studio/bin/unsloth"

# Use devnull for output or let it log to stdout so it does not block the buffer
process = subprocess.Popen(
    [unsloth_path, "studio", "-H", "0.0.0.0", "-p", str(PORT)],
    stdout=None,
    stderr=None
)

# Keep cell active
try:
    while True:
        time.sleep(1)
except KeyboardInterrupt:
    print("Stopping server...")
    process.terminate()
```
